Classify movies based on ratings

```python
def rating_group(rating):
  if rating >= 7.5:
    return "Good"
  elif rating >= 6.0:
    return "Average"
  else:
    return "Bad"
```

```python
df["Rating_Category"] = df["Rating"].apply(rating_group)
```
